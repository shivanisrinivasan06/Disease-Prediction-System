# Disease Prediction System

## Project Overview

This project uses machine learning to classify whether a patient belongs
to the No Disease (0) or Disease (1) class based on health-related features.

## Dataset
- UCI Heart Disease Dataset
- 303 records
- 13 features

## Machine Learning Models
1. Logistic Regression
2. Decision Tree

## Best Model
Logistic Regression

## Best Accuracy
86.89%

## Conclusion

The Disease Prediction System was developed using the UCI Heart Disease
dataset. The data was cleaned by handling missing values, followed by
exploratory data analysis and preparation of a binary target variable.

Two classification algorithms were tested: Logistic Regression and
Decision Tree. Logistic Regression achieved an accuracy of 86.89%,
while Decision Tree achieved 73.77%. Therefore, Logistic Regression
was selected as the better-performing model for this project.

The final model was also tested with new patient data and was saved
along with the feature scaler for future use.

This project is intended for educational purposes and should not be
used as a medical diagnostic system.

```python
y_binary = y_binary["num"]

print(y_binary.head())
print(y_binary.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_binary,
    test_size=0.2,
    random_state=42,
    stratify=y_binary
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
print(X_train_scaled.shape)
print(X_test_scaled.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(random_state=42)

model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test_scaled)

print(y_pred)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy percentage:", accuracy * 100)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print(cm)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```

```python
print(classification_report(y_test, y_pred))
```

```python
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(random_state=42)

dt_model.fit(X_train, y_train)
```

```python
dt_pred = dt_model.predict(X_test)

print(dt_pred)
```

```python
from sklearn.metrics import accuracy_score

dt_accuracy = accuracy_score(y_test, dt_pred)

print("Decision Tree Accuracy:", dt_accuracy)
print("Decision Tree Accuracy Percentage:", dt_accuracy * 100)
```

```python
print("Logistic Regression Accuracy:", accuracy)
print("Decision Tree Accuracy:", dt_accuracy)
```

```python
model_comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "Decision Tree"],
    "Accuracy": [accuracy, dt_accuracy]
})

model_comparison
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(y_test, y_pred)

plt.title("Logistic Regression - Confusion Matrix")
plt.show()
```
